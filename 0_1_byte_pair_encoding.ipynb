```python
class ByteTokenizer():
  def __init__(self) -> None:
    pass

  def encode(self, x):
    return [y for y in bytes(x, encoding='utf-8')]

  def decode(self, x):
    return ''.join([chr(y) for y in x])
```

```python
t = ByteTokenizer()
```

```python
text = 'Hi, I am Viswajith. How are you doing?'
len(text)
```

```python
enc = t.encode(text)
len(enc), enc
```

```python
ori = t.decode(enc)
ori
```
